```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Generate initial data
np.random.seed(42)
X = 2 * np.random.rand(10, 1)  # 10 random points between 0 and 2
y = 4 + 3 * X + np.random.randn(10, 1)  # y = 4 + 3x + noise

# Initialize parameters
w = np.random.randn(1)
b = np.random.randn(1)

# Gradient Descent Function
def gradient_descent(X, y, w, b, alpha, epochs):
    m = len(X)
    for epoch in range(epochs):
        # Compute predictions
        predictions = w * X + b

        # Compute gradients
        dw = -(1 / m) * np.sum(X * (y - predictions))
        db = -(1 / m) * np.sum(y - predictions)

        # Update parameters
        w = w - alpha * dw
        b = b - alpha * db

    return w, b

# Loss Function (Mean Squared Error)
def compute_loss(X, y, w, b):
    m = len(X)
    predictions = w * X + b
    loss = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    return loss

# Function to update the plot
def update_plot(learning_rate, epochs, num_points):
    global w, b
    w = np.random.randn(1)  # Reset parameters
    b = np.random.randn(1)

    # Use the first `num_points` points
    X_subset = X[:num_points]
    y_subset = y[:num_points]

    # Run Gradient Descent
    w, b = gradient_descent(X_subset, y_subset, w, b, learning_rate, epochs)

    # Clear the previous plot
    plt.clf()

    # Plot the data points
    plt.scatter(X_subset, y_subset, color='blue', label="Data Points")

    # Plot the regression line
    x_vals = np.linspace(0, 2, 100)
    y_vals = w * x_vals + b
    plt.plot(x_vals, y_vals, color='red', label="Regression Line")

    # Add labels and title
    plt.xlabel("X")
    plt.ylabel("y")
    plt.title(f"Gradient Descent (Learning Rate: {learning_rate}, Epochs: {epochs}, Points: {num_points})")
    plt.legend()
    plt.xlim(0, 2)
    plt.ylim(0, 12)
    plt.show()

    # Print the final parameters and loss
    loss = compute_loss(X_subset, y_subset, w, b)
    print(f"Final Parameters: w = {w[0]:.2f}, b = {b[0]:.2f}")
    print(f"Final Loss: {loss:.2f}")

# Create widgets for dynamic control
learning_rate_slider = widgets.FloatSlider(value=0.1, min=0.01, max=1.0, step=0.01, description='Learning Rate:')
epochs_slider = widgets.IntSlider(value=50, min=1, max=200, step=1, description='Epochs:')
points_slider = widgets.IntSlider(value=5, min=1, max=10, step=1, description='Number of Points:')

# Link widgets to the update function
widgets.interact(update_plot,
                 learning_rate=learning_rate_slider,
                 epochs=epochs_slider,
                 num_points=points_slider)
```
